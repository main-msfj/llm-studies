```python
# Attached through `dataset_sources` in kernel-metadata.json, so Kaggle mounts it
# under /kaggle/input and nothing is downloaded. It is too big to pull locally.
DATA_HANDLE = "arghadeeppakhira/125m-llm-dataset"

if not ON_KAGGLE:
    raise RuntimeError(f"This notebook runs on Kaggle: {DATA_HANDLE} is not downloaded locally")

TRAIN_FILE = next(Path("/kaggle/input").rglob("train_small.txt"))
DATA_DIR = TRAIN_FILE.parent
TOKENIZER_MODEL = DATA_DIR / "mini.model"  # SentencePiece model shipped with the dataset

print(f"Data dir: {DATA_DIR}")
for f in sorted(DATA_DIR.iterdir()):
    print(f"  {f.name:<20} {f.stat().st_size / 1e6:>10.1f} MB")
```
